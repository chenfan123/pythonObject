```python
from enum import Enum

from pydantic import BaseModel, Field


class MessageRole(str, Enum):
    USER = "user"
    ASSISTANT = "assistant"
    SYSTEM = "system"


class MessageDTO(BaseModel):
    role: MessageRole
    content: str = Field(min_length=1)

# MessageRole 限制角色只能是 user, assistant, system
```

继承 str, Enum 的好处是：
- 在 Python 中具有枚举语义；
- JSON 中表现为字符串；
- OpenAPI 能展示允许的值。

### 嵌套模型

```python
class ConversationCreateDTO(BaseModel):
    title: str = Field(min_length=2, max_length=100)

    messages: list[MessageDTO] = Field(
        min_length=1,
        max_length=100,
    ) # messages 必须是列表,列表中的每个元素都必须符合 MessageDTO

    metadata: dict[str, str] = Field(
        default_factory=dict,
    )
```

```python
conversation = ConversationCreateDTO.model_validate(
    {
        "title": "Python 学习",
        "messages": [
            {
                "role": "user",
                "content": "解释装饰器",
            }
        ],
    }
)
```

### 可选嵌套模型

```python
class AttachmentDTO(BaseModel):
    filename: str
    url: str


class MessageDTO(BaseModel):
    role: MessageRole
    content: str = Field(min_length=1)
    attachment: AttachmentDTO | None = None
```

### 字典约束

```python
metadata: dict[str, str] # 要求键和值都是字符串。

from typing import Any

metadata2: dict[str, Any] # 允许任意值，但也意味着 Pydantic 对值的约束大幅降低
```

### 列表、集合与元组

```python
tags: list[str] # 保留顺序也允许重复
tags: set[str] # 不保留顺序，也不允许重复
coordinates: tuple[float, float] # 固定长度元组
# JSON 中没有原生 set 或 tuple，序列化时通常仍表现为数组。
```

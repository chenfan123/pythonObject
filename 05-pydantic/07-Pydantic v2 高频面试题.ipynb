##### Pydantic 解决什么问题？
> Pydantic 根据 Python 类型注解，对外部输入进行解析、转换和校验，并将其转换为结构明确的模型对象。它还支持序列化、JSON Schema、字段别名和自定义验证器，经常用于 FastAPI 的请求与响应边界。

> 关键词：解析 + 转换 + 校验 + 序列化

##### model_validate() 和构造函数有什么区别？
区别主要体现在表达意图：
`UserDTO(username="Alice")`看起来像使用参数创建 Python 对象。

`UserDTO.model_validate(data)`使用一份外部数据校验并构建模型。

##### 字段验证器和模型验证器有什么区别？
**字段验证器**
处理单个字段，适合：
- 去除空格
- 转换大小写
- 检查单字段格式
- 检查单字段业务无关约束

**模型验证器**
处理多个字段之间的关系，适合：
- 两次密码是否一致
- 开始时间是否早于结束时间
- 至少提供两个字段中的一个
- 字段组合是否合法

##### model_dump() 和 model_dump_json() 有什么区别？
第一个返回的是 Python 对象

第二个返回的是已经编码的 JSON 字符串

##### model.model_dump中的exclude参数有什么用？
- exclude_none=True ：排除值为 None 的字段
- exclude_defaults=True ：排除值与默认值相同的字段
- exclude_unset=True ：排除值未发生改变的字段

> none 看当前值，unset 看用户有没有传，defaults 看当前值是否等于默认值。

##### 为什么要区分请求模型和响应模型？
> 请求模型定义客户端允许提交的数据，响应模型定义服务端允许公开的数据。分离两者可以避免密码、成本价和内部状态等字段泄漏，也能让输入与输出独立演进。

## HTTP 请求
一个请求包含：
- 请求方法
- URL
- 请求头
- 请求体
```
POST /users?notify=true HTTP/1.1
Host: api.example.com
Authorization: Bearer token
Content-Type: application/json

{
  "name": "Alice",
  "email": "alice@example.com"
}
```
## HTTP 响应
- 状态码
- 响应头
- 响应体
```
HTTP/1.1 201 Created
Content-Type: application/json

{
  "id": 100,
  "name": "Alice",
  "email": "alice@example.com"
}
```

## 常见 HTTP 方法
- GET：获取资源
- POST：创建资源
- PUT：完整替换资源
- PATCH：部分更新资源
- DELETE：删除资源
- OPTIONS：获取资源支持的 HTTP 方法
- HEAD：获取资源头
- TRACE：回显请求头
- CONNECT：建立隧道

## 常见客户端错误
```
400 Bad Request   请求格式或业务参数明显错误
401 Unauthorized  未登录或凭证无效
403 Forbidden     已识别身份，但没有权限
404 Not Found     资源不存在
409 Conflict      资源状态冲突
422 Unprocessable Entity  数据格式可解析，但校验失败
429 Too Many Requests     请求过多
```

## 常见服务端错误
```
500 Internal Server Error  服务内部未知错误
502 Bad Gateway            上游服务返回无效响应
503 Service Unavailable    服务暂时不可用
504 Gateway Timeout        等待上游服务超时
```

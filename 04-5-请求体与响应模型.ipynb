FastAPI 通常使用 Pydantic Model 描述请求体。

```python
from pydantic import BaseModel, Field


class UserCreate(BaseModel):
    name: str = Field(min_length=2, max_length=50)
    email: str
    password: str = Field(min_length=8)
```

```python
from fastapi import FastAPI, status
app = FastAPI()

@app.post(
    "/users",
    status_code=status.HTTP_201_CREATED,
)
async def create_user(payload: UserCreate):
    return payload
```

读取 JSON Body
→ 解析 JSON
→ 使用 UserCreate 校验
→ 创建 UserCreate 实例
→ 传给 create_user

Pydantic v2 中转换为字典。`payload.model_dump()`

```python
class UserCreate(BaseModel):
    name: str = Field(min_length=2, max_length=50)
    email: str
    password: str = Field(min_length=8)


class UserResponse(BaseModel):
    id: int
    name: str
    email: str


@app.post(
    "/users",
    response_model=UserResponse,
    status_code=status.HTTP_201_CREATED,
)
async def create_user(payload: UserCreate):
    created_user = {
        "id": 1,
        "name": payload.name,
        "email": payload.email,
        "password": payload.password,
        "internal_note": "new user",
    }

    return created_user
```

#### response_model 的安全价值
使用响应模型建立明确的输出边界

##### 请求校验失败与响应校验失败
请求数据不符合 UserCreate： 422
服务端返回的数据不符合 UserResponse：500

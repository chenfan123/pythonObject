### 字符串约束


```python
class UserDTO(BaseModel):
    username: str = Field(
        min_length=3,
        max_length=20,
        pattern=r"^[a-zA-Z0-9_]+$",
    )
```

### 数值约束

```python
class ProductDTO(BaseModel):
    price: float = Field(gt=0)
    stock: int = Field(ge=0)
    discount: float = Field(ge=0, le=1)
    quantity: int = Field(multiple_of=5) # 必须5的倍数
```

### 集合约束

```python
class ArticleDTO(BaseModel):
    tags: list[str] = Field(
        min_length=1, # 至少一个标签
        max_length=5, # 最多5个标签
    )
```

### 必填字段

```python
class UserDTO(BaseModel):
    name: str # 没有默认值就是必填字段
```

### 有默认值的字段

```python
class UserDTO(BaseModel):
    active: bool = True
```

### 可空不等于可省略

```python
nickname: str | None # 字段必填 但值可以是字符串或 null

nickname: str | None = None # 字段可以省略
```

### default_factory

```python
# 列表默认值推荐
class ArticleDTO(BaseModel):
    tags: list[str] = Field(default_factory=list)

# 每个模型实例都会获得自己的新列表
```

```python
# 动态时间默认值
from datetime import datetime, timezone


class EventDTO(BaseModel):
    created_at: datetime = Field(
        default_factory=lambda: datetime.now(timezone.utc)
    ) # 每次创建实例时都会重新计算当前时间
```

`default_factory=list`
传入的是函数对象 list，Pydantic 创建每个实例时调用`list()`，清楚地表达“每次创建新默认值”

# PostgreSQL 学习环境

## 启动数据库

在 `07-postgresql` 目录执行：

```bash
docker compose up -d
```

## 连接信息

| 项 | 值 |
|---|---|
| Host | `localhost` |
| Port | `5432` |
| User | `learn` |
| Password | `learn` |
| Database | `learn_db` |
| URL | `postgresql+psycopg://learn:learn@localhost:5432/learn_db` |

后续每趴 notebook 开头都先跑「连接」那一格。

````python
%load_ext sql
%config SqlMagic.displaylimit = 50

%sql postgresql+psycopg://learn:learn@localhost:5432/learn_db
````

````python
%%sql
SELECT current_user, current_database(), version();
````

看到用户 `learn`、库 `learn_db` 就说明环境就绪。

常用命令：

```bash
docker compose up -d      # 启动
docker compose stop       # 停止（数据保留）
docker compose down       # 停并删容器（数据卷默认还在）
docker compose down -v    # 连数据一起清空（慎用）
```

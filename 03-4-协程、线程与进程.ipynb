### 并发与并行
并发是多个任务在同一段时间内推进：可以交替执行，不要求同一时刻真的运行。
并行是多个任务在同一时刻执行。

### 协程
协程通常运行在单个事件循环线程中。
- 由事件循环调度
- 通常是协作式切换
- 在可暂停的await处让出执行权
- 创建和切换成本较低
- 多个协程通常共享同一线程和内存
- 适合大量I/O密集型任务

常见场景：
- 异步HTTP请求
- 异步数据库查询
- redis请求
- websocket
- sse流式输出
- llm api调用

不适合CPU密集计算

### 线程
线程由操作系统调度，一个进程可以包含多个线程。
- 多个线程共享进程内存；
- 可以运行同步阻塞函数；
- 切换成本高于协程；
- 共享数据可能产生竞态条件；
- 需要锁等同步机制；
- 在典型 CPython 中受 GIL 影响。
  
适合场景：
- 同步文件操作；
- 同步 HTTP SDK；
- 同步数据库驱动；
- 无法改成原生异步的阻塞 I/O；
- 大量时间用于等待外部资源的同步函数。

### 进程
进程拥有独立的：
- python解释器
- 内存空间
- GIL
- 系统资源

多个进程可以利用多个CPU核心，实现并行计算。

适合场景：
- 纯 Python CPU 密集计算
- 需要大量系统资源的任务
- 需要隔离解释器和内存空间的应用
- 需要并行计算的场景

> 协程适合大量原生异步 I/O，通过事件循环在 await 处协作式切换；线程适合同步阻塞 I/O，可以避免阻塞事件循环，但共享内存会带来线程安全问题，且纯 Python CPU 计算受 GIL 影响；进程拥有独立解释器和内存，可利用多核并行，适合 CPU 密集任务，但创建、通信和序列化成本更高。

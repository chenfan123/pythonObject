````python
from typing import Annotated

from fastapi import Cookie, FastAPI, Header, Path, Query


app = FastAPI()


@app.get("/users/{user_id}")
async def get_user(
    user_id: Annotated[int, Path(gt=0)],
    keyword: Annotated[
        str | None,
        Query(min_length=2, max_length=50),
    ] = None,
    page: Annotated[int, Query(ge=1)] = 1,
    user_agent: Annotated[str | None, Header()] = None,
    session_id: Annotated[str | None, Cookie()] = None,
):
    return {
        "user_id": user_id,
        "keyword": keyword,
        "page": page,
        "user_agent": user_agent,
        "session_id": session_id,
    }
````

### Path 参数
Path 参数一定是必填的，因为没有它就无法匹配该路由.

路由中使用大括号声明`@app.get("/users/{user_id}")`，函数中使用同名参数接收。

### Query 参数
URL 问号后面的参数是 Query `/users?keyword=python&page=2`
#### Query 参数别名
外部 API 可能使用驼峰命名,Python 使用下划线,可以添加别名:
```
sort_by: Annotated[
    str,
    Query(alias="sortBy"),
] = "created_at"
```
#### 多值 Query 参数
可以接受为列表`tags: Annotated[list[str] | None, Query()] = None`

#### 默认值
可以设置默认值`page: Annotated[int, Query(ge=1)] = 1`


### Header参数
默认情况下，FastAPI 会把 Python 参数名中的下划线转换为 HTTP Header 的连字符

`x_request_id: Annotated[str | None, Header()] = None`
对应`X-Request-Id: request-123`

### Cookie 参数
`session_id: Annotated[str | None, Cookie()] = None`读取`Cookie: session_id=abc123`

### 如何判断来源
名称出现在 /path/{name} 中
→ Path

显式使用 Query()
→ Query

显式使用 Header()
→ Header

显式使用 Cookie()
→ Cookie

简单类型且不在 Path 中
→ 通常默认为 Query

Pydantic Model
→ 通常作为 Body

# asyncio.wait_for()

```python
import asyncio

async def worker():
    print("start")
    try:
        await asyncio.sleep(2)
        print("end")
        return "success"
    except asyncio.CancelledError:
        print("cancelled")
        raise
    finally:
        print("cleanup")

async def main():
    try:
        result = await asyncio.wait_for(
            worker(),
            timeout=1,
        )
        print(result)
    except asyncio.TimeoutError:
        print("timeout")
    print("after")

asyncio.run(main())
# worker 输出 start → sleep(2) 开始等待 → 一秒后 wait_for 超时 → worker 在 await 处收到 CancelledError → except 输出 cancelled，并重新 raise → finally 输出 cleanup → wait_for 对外表现为超时 → main 捕获超时并输出 timeout → main 继续输出 after
```

表示最多等待一秒。
如果任务在一秒内没有完成：
1. wait_for() 请求取消内部任务；
2. 协程会在暂停点收到 asyncio.CancelledError；
3. 清理逻辑执行；
4. 调用方收到超时异常。

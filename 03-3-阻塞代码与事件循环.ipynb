```python

import time
import asyncio

async def bad_job():
    time.sleep(2)
    await asyncio.sleep(2)
```

```python
async def blocking_job():
    time.sleep(2)
```

time.sleep同步阻塞
    - 事件循环无法调度其他协程
    - 其他网络请求、定时器和 Task 都无法推进
asyncio.sleep异步等待
    - 当前协程暂停，控制权返回事件循环
    - 其他 Task 可以继续运行

async def 只允许函数使用异步机制，不会自动改造内部的同步代码

### 处理同步阻塞 I/O
如果必须调用同步阻塞函数，可以将它放入线程

```python
def read_file_sync():
    time.sleep(2)
    return "data"


async def main():
    data = await asyncio.to_thread(read_file_sync)
    print(data)
```

to_thread() 主要适合：

- 同步文件操作；m
- 不支持异步的第三方 HTTP SDK；
- 阻塞式数据库驱动；
- 其他 I/O 密集型同步函数。

线程切换也有成本，不应把所有普通函数都放进线程

对于纯 Python CPU 密集计算，通常应考虑进程池，而不是指望 asyncio 提升并行计算速度

### CPU 密集任务
> 面试模版：原生异步 I/O 使用 await；不得不调用同步阻塞 I/O 时使用线程池或 asyncio.to_thread()；纯 Python CPU 密集任务通常交给进程池或独立 Worker，避免阻塞事件循环

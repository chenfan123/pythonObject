````python
from fastapi import FastAPI


app = FastAPI()


@app.get("/health")
async def health():
    return {"status": "ok"}
````

3个重要组合
### 1. fastapi应用对象
`app = FastAPI()`
它负责：
- 保存路由；
- 接收 HTTP 请求；
- 匹配对应处理函数；
- 执行参数解析和校验；
- 生成 HTTP 响应；
- 生成 OpenAPI 文档。
```
# 可以给应用配置元数据，会出现在API文档中
app = FastAPI(
    title="Agent Backend",
    description="Agent 后端学习项目",
    version="0.1.0",
)
```
### 2. 注册路由操作
```
@app.get("/health")
async def health():
    return {"status": "ok"}
```
`@app.get("/health")` 是注册路由操作，它将 `health` 函数注册为处理 `/health` 路径的 GET 请求。

### 3. 返回字典


`return {"status": "ok"}` 是返回字典操作，它将返回一个字典对象。生成 HTTP 响应，默认成功状态码

## 装饰器什么时候执行？
应用启动并导入模块
→ 创建 app
→ 定义 health
→ 把 GET /health 与 health 关联
→ 等待请求

只有收到对应 HTTP 请求后，才执行 health()

## Uvicorn 的作用
FastAPI 负责应用逻辑，但它本身不直接监听网络端口。
Uvicorn 是 ASGI Server，负责：
- 监听 IP 和端口；
- 接收客户端连接；
- 解析 HTTP 请求；
- 调用 FastAPI 应用；
- 把 FastAPI 的响应发送给客户端。
  
## 启动 FastAPI 应用
如果py文件名为main.py，app = FastAPI()
```
uvicorn main:app --reload

# main → 导入 main.py 模块
# app  → 从模块中找到名为 app 的对象
# --reload 会监听代码变化并自动重启服务，适合开发环境。
```
